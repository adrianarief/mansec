# Data Tranformation & cleansing

```python
import pandas as pd
from datetime import datetime
```

## Import Data

```python
file_path1 = "C:/Users/adria/Documents/Test/Mandiri Sekuritas/users_data.csv" 

file_path2 = "C:/Users/adria/Documents/Test/Mandiri Sekuritas/cards_data.csv"  

file_path3 = "C:/Users/adria/Documents/Test/Mandiri Sekuritas/transactions_data.csv"  

df_users = pd.read_csv(file_path1)
df_cards = pd.read_csv(file_path2)
df_trx = pd.read_csv(file_path3)
```

## Initial Data Users

```python
df_users.head()
# df_users.info()
```

## Change Users data with '$' to Decimal Number

```python
# Remove the '$' sign from the 'amount' column
df_users['per_capita_income'] = df_users['per_capita_income'].str.replace('$', '', regex=False)

df_users['yearly_income'] = df_users['yearly_income'].str.replace('$', '', regex=False)

df_users['total_debt'] = df_users['total_debt'].str.replace('$', '', regex=False)

df_users.head()
```

## Initial Data Cards

```python
df_cards.head()
# df_cards.info()
```

## Change Cards data with '$' to Decimal Number

```python
# Remove the '$' sign from the 'amount' column
df_cards['credit_limit'] = df_cards['credit_limit'].str.replace('$', '', regex=False)

df_cards.head()
```

## Initial Trx Data

```python
df_trx.head()
# df_trx.info()
```

## Change Trx data with '$' to Decimal Number

```python
# Remove the '$' sign from the 'amount' column
df_trx['amount'] = df_trx['amount'].str.replace('$', '', regex=False)

df_trx.head()
```

## Made the data to csv file

```python
df_users.to_csv('Dim Users')
df_cards.to_csv('Dim Cards')
df_trx.to_csv('Fact Trx')
```
